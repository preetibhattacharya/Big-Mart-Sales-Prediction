```python
# Step 1: Build canonical sensor targets from Excel + YAML
import re, json
import pandas as pd
import yaml
from pathlib import Path

# --- adjust these if your variable names/paths differ ---
EXCEL_PATH = "/content/Stand13-18_filtered_signals.xlsx"   # or use excel_df already in memory
YAML_PATH = "stand13_18.yaml"
OUT_TARGETS = "/content/targets.json"

# helper normalizer: remove brackets/quotes/whitespace, keep digits, colon, dot, dash
def normalize_sensor_id(s):
    if s is None: return ''
    s = str(s)
    s = re.sub(r'[\[\]\s\'"]', '', s)          # remove brackets and quotes and spaces
    s = s.strip()
    # sometimes ids use comma separation, keep only first token
    s = s.split(',')[0]
    return s

# 1. Load excel (if excel_df exists in notebook, you can skip reading again)
try:
    excel_df  # if variable exists, use it
    print("Using excel_df already in memory.")
except NameError:
    excel_df = pd.read_excel(EXCEL_PATH)
    print(f"Read excel from {EXCEL_PATH}")

# check common columns
print("Excel columns:", excel_df.columns.tolist())

# Attempt to detect the sensor id column name
possible_cols = [c for c in excel_df.columns if 'sensor' in c.lower() or 'id' in c.lower()]
sensor_col = possible_cols[0] if possible_cols else excel_df.columns[0]
print("Using Excel sensor column:", sensor_col)

# Normalize Excel sensor ids
excel_df['sensor_norm'] = excel_df[sensor_col].astype(str).apply(normalize_sensor_id)

# Load YAML
with open(YAML_PATH) as f:
    yaml_data = yaml.safe_load(f)

# Extract ids from YAML structure (control_signal + affected_signals)
yaml_ids = []
equip = yaml_data.get('Equipment', {}) if isinstance(yaml_data, dict) else {}
for comp, info in equip.items():
    if not isinstance(info, dict):
        continue
    # control_signal may be a string
    cs = info.get('control_signal')
    if cs:
        yaml_ids.append(normalize_sensor_id(cs))
    # affected_signals may be a list
    for k in ('affected_signals','unaffected_signals'):
        vals = info.get(k)
        if isinstance(vals, list):
            yaml_ids += [normalize_sensor_id(v) for v in vals]
        elif vals:
            yaml_ids.append(normalize_sensor_id(vals))

# Build combined target set
excel_ids = [s for s in excel_df['sensor_norm'].unique() if s and s.lower() != 'nan']
yaml_ids = [s for s in yaml_ids if s and s.lower() != 'nan']

targets_set = set(excel_ids) | set(yaml_ids)
targets = sorted(targets_set)

# Save to JSON for later use
Path(OUT_TARGETS).parent.mkdir(parents=True, exist_ok=True)
with open(OUT_TARGETS, 'w') as f:
    json.dump(targets, f, indent=2)

# Print summary for you to copy back
print("\n--- TARGETS SUMMARY ---")
print("Number of targets (unique):", len(targets))
print("First 30 targets sample:", targets[:30])
print("\nNumber of excel-only ids:", len(set(excel_ids) - set(yaml_ids)))
print("Number of yaml-only ids:", len(set(yaml_ids) - set(excel_ids)))
print("Number in both:", len(set(excel_ids).intersection(yaml_ids)))

# show any odd-looking ids (non-digit characters besides : . -)
odd = [s for s in targets if re.search(r'[^0-9:.\-]', s)]
if odd:
    print("\nWARNING: Found odd-formatted IDs (check these):", odd[:30])

# Save small preview CSV for inspection
pd.DataFrame({'target_id': targets}).head(200).to_csv('/content/targets_preview.csv', index=False)
print("\nSaved /content/targets.json and /content/targets_preview.csv")
```

```python
import pandas as pd
import glob

def extract_and_save(parquet_files, day_label, chunk_size=50):

    buffer = []
    chunk_id = 0

    for i, fp in enumerate(parquet_files):

        df = pd.read_parquet(fp)
        df.columns = [normalize_sensor(c) for c in df.columns]

        df_target = df[final_targets].copy()
        df_target["timestamp"] = df.index
        df_target["source_file"] = fp

        buffer.append(df_target)

        # save chunk
        if (i+1) % chunk_size == 0:
            master = pd.concat(buffer, ignore_index=True)
            out_path = f"{SAVE_DIR}/{day_label}_chunk_{chunk_id}.csv"
            master.to_csv(out_path, index=False)

            print(f" Saved {out_path} | rows: {master.shape[0]}")

            buffer = []
            chunk_id += 1

    # save remainder
    if len(buffer) > 0:
        master = pd.concat(buffer, ignore_index=True)
        out_path = f"{SAVE_DIR}/{day_label}_chunk_{chunk_id}.csv"
        master.to_csv(out_path, index=False)

        print(f" Saved FINAL {out_path} | rows: {master.shape[0]}")
```
